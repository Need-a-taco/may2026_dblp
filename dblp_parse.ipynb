```python
import pandas as pd
```

```python
group1 = [
    "Large Language Models",
    "Generative Artificial Intelligence",
    "LLM",
    "LLMs",
    "large language model",
    "foundation model",
    "generative artificial intelligence",
    "generative pre-trained transformer",
    "generative AI",
    "chatbot",
    "ChatGPT",
    "GPT-4",
    "GPT-5",
    "deepseek",
    "gemini",
]

group2 = [
    "Health Equity",
    "Healthcare Disparities",
    "Bias, Implicit",
    "Prejudice",
    "Racism",
    "Systemic Racism",
    "Social Discrimination",
    "bias",
    "debias",
    "fair",
    "inequ" ,
    "disparit" ,
    "equit",
    "sociodemographic",
    "demographic",
    "race", 
    "racial", 
    "racism",
    "ethnic",
    "gender", 
    "sex",
    "sexism",
    "ageism",
    "stereotype",
    "underserved", 
    "underrepresented", 
    "marginalized",
]

group3 = [
    "Delivery of Health Care",
    "Health Care Sector",
    "Medicine",
    "Health Services",
    "health",
    "clinic",
    "medicine",
    "medical",
    "biomedical",
    "patient",
    "hospital",
]

def query(text):
    if not text:
        return False  
    text = str(text).lower()
    
    for group in [group1, group2, group3]:
        if not any(term.lower() in text for term in group):
            return False     
    return True
```

```python
results = []

chunk_count = 0

for df in pd.read_json(
    "dblp.jsonl",
    lines=True,
    chunksize=100000
):

    chunk_count += 1

    print("Processing chunk:", chunk_count)

    df["tiabs"] = (
        df["title"].fillna("").astype(str)
    )

    new_filtered_title_df = df[df['tiabs'].apply(query)]

    new_filtered_title_df = new_filtered_title_df[
        ["title", "author", "year"]
    ]

    results.append(new_filtered_title_df)


final_df = pd.concat(results)

final_df.to_csv("new_dblp_query.csv", index=False)

print("Done.")
print("Total matches:", len(final_df))
```
